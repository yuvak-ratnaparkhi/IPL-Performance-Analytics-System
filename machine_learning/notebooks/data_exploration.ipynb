### 1. Top Winning Teams

```python
plt.figure(figsize=(12,6))

df["winner"].value_counts().head(10).plot(kind="bar")

plt.title("Top Winning Teams")
plt.xlabel("Team")
plt.ylabel("Wins")

plt.show()
```

This chart shows the teams with the highest number of match victories in IPL history.

### 2. Matches Per Season

```python
plt.figure(figsize=(12,5))

df["season"].value_counts().sort_index().plot(
    kind="line",
    marker="o"
)

plt.title("Matches Per Season")
plt.xlabel("Season")
plt.ylabel("Matches")

plt.show()
```

This chart displays the number of IPL matches played in each season.

### 3. Top Venues

```python
plt.figure(figsize=(12,6))

df["venue"].value_counts().head(10).plot(
    kind="bar"
)

plt.title("Top IPL Venues")
plt.xlabel("Venue")
plt.ylabel("Matches Hosted")

plt.show()
```

This chart shows the venues that have hosted the most IPL matches.

 ### 4. Toss Decision Analysis

```python
plt.figure(figsize=(6,4))

sns.countplot(
    x="toss_decision",
    data=df
)

plt.title("Toss Decision Distribution")

plt.show()
```

This chart compares how often captains chose to bat or field after winning the toss.

### 5. Team Appearance Frequency

```python
all_teams = pd.concat([
    df["team1"],
    df["team2"]
])

plt.figure(figsize=(12,6))

all_teams.value_counts().plot(kind="bar")

plt.title("Team Appearance Frequency")
plt.xlabel("Team")
plt.ylabel("Matches Played")

plt.show()
```

This chart shows how many matches each team has participated in across IPL seasons.
